```python
month = int(input("Enter month number (1-12): "))

months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

if month < 1 or month > 12:
    print("Invalid month number")
elif month == 2:
    print(f"{months[month-1]} is a 28-day month.")
elif month in [4, 6, 9, 11]:
    print(f"{months[month-1]} is a 30-day month.")
else:
    print(f"{months[month-1]} is a 31-day month.")
```

```python
val =  input("Enter char : ")

if len(val) == 1:

    if (val == 'a' or val == 'e' or val == 'i' or val == 'o' or val == 'u') or (val == 'A' or val == 'E'):
        print(val, "is vowal")
    elif 'a' <= val <= 'z' or 'A' <= val <= 'z':
        print(val, "is consonant")
    else:
        print(val,"is special charater")
else:
    print("Not a charater")
        
```
