1. WAP to check if numbers are divisible by 4 and 5.

input: num1 = 20
output:
15 is divisible by 4 and 5


input: num1 = 15
output:
15 is not divisible by 4 and 5


```python
num1 = int(input("Enter num1 : "))

if num1 % 4 == 0 and num1 % 5 == 0:
    print(f"{num1} is divisible by 4 and 5")
else:
    print(f"{num1} is not divisible by 4 and 5")
   
```

2. WAP to determine whether the entered angles define a right-angled triangle.Take three values of angle from the user.



```python
angle1 = int(input("Enter angle1: "))
angle2 = int(input("Enter angle2: "))
angle3 = int(input("Enter angle3: "))

if (angle1 + angle2 + angle3) == 180:
    print("it is a right-angle triangle")
else:  
    print("it is not a right-angle triangle")

```
